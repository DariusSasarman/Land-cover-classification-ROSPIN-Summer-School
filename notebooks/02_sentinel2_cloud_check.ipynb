```python
import rasterio
import matplotlib.pyplot as plt
import numpy as np

with rasterio.open("../data/raw/sentinel2_aoi.tif") as src:
    print("Number of bands:", src.count)
    print("Size (width x height):", src.width, src.height)
    print("CRS:", src.crs)

    red = src.read(3)
    green = src.read(2)
    blue = src.read(1)

rgb = np.dstack([red, green, blue]).astype(float)
rgb = np.clip(rgb / np.percentile(rgb, 98), 0, 1)  # normalizare pentru afisare corecta

plt.figure(figsize=(10, 10))
plt.imshow(rgb)
plt.title("Sentinel-2 - Cluj AOI")
plt.axis('off')
plt.show()
```
