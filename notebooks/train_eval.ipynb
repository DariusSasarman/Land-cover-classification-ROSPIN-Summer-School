```python
# Anonymous Helper Functions

# Random haze application via custom transformation
haze_transform = lambda img: color_cast_approx(  

    img, 
    haze_intensity=max_haze,
    fixed_alpha=False,

)  

# Non-blur/non-interpolation implementation of D4 symmetry rotation
rotational_transform = lambda img: torch.rot90(

    img, 
    k=int(torch.randint(0, 4, ()).item()),  # `torch.rot90()` expects `k` as Python integer
    dims=(-2, -1),

)  

# Scaling setup to enable cropping of black areas introduced by rotational transformation
#   (!) Do not use: Blurring due to interpolation does destroy features at low res (64x64)
# final_dim = 64
# scaling_factor = 1.5  # conservative
# scale_dim = int(final_dim * scaling_factor)

# Transformations (Data Augmentation)
#   - Reference: https://docs.pytorch.org/vision/main/auto_examples/transforms/plot_transforms_illustrations.html#sphx-glr-auto-examples-transforms-plot-transforms-illustrations-py

train_transformations = v2.Compose(  # Using `torchvision.transforms.v2` instead of `torchvision.transforms`

    [

        # (!) `v2.ToTensor()` depreaceatd in "v2" -> use `v2.ToImage()` ... `v2.ToDtype(torch.float32, scale=True)` instead

        v2.ToImage(),  # Convert to PIL Image or Tensor ()

        v2.Resize((64, 64), antialias=True),  # For consistency as EurosSAT dimensions already 64x64
        # v2.Resize((scale_dim, scale_dim), antialias=True),

        # (!) `v2.RandomRotation` can introduce blurring/anti-aliasing -> avoid here
        # v2.RandomApply([v2.RandomRotation(90)], p=0.5),  # prev: [-degrees, +degrees] rotations
        v2.Lambda(rotational_transform),  # wrapped via `v2.Lambda()` for consistency

        # Reasonable spatial variation
        v2.RandomHorizontalFlip(p=0.5),
        # v2.RandomVerticalFlip(p=0.5),

        # v2.RandomResizedCrop(  # Scale/positon variation robustness

        #     size=(64, 64), 
        #     scale=(0.85, 1.0), 
        #     ratio=(0.95, 1.05), 
        #     antialias=True

        # ),  

        # Theoretically sound selection
        # v2.RandomCrop(size=(final_dim, final_dim)),
        # v2.gaussian_blur(kernel_size=(5, 9), sigma=(0.1, 1.0)),  # Not used: already very low resolution
        # v2.RandomAffine(degrees=0, translate=(0.05, 0.05), scale=(0.95, 1.05))

        # (!) Convert after geometric transformations / before color transforms
        #   - Gemoetric trans. more efficient/stable operating on `uint8` tensors
        #   - Color trans. require floating-point tensor in [0, 1] range (`scale=True`)
        v2.ToDtype(torch.float32, scale=True),

        # ColorJitter: Tuned for realistic examples (manifesting atmospheric effects)
        #   - Conservative approach: Classes are partially color/texture differentiated  
        v2.RandomApply(

            [

                # Approx. atmospheric/lighting condition variance
                #   - Sentinel-2 (EuroSAT data source): Assume consitent calibration
                v2.ColorJitter(  # Conservative

                    brightness=0.15,  # prev: 0.20
                    contrast=0.15,  # prev: 0.20
                    saturation=0.10,  # prev: 0.15
                    hue=0.00,  # Hue shifts do not correspond to realistic atmospheric effects

                )

            ], p=0.35,  # prev: 0.50

        ),  

        v2.RandomApply(

            # Custom Color Cast Approximation
            [v2.Lambda(haze_transform)], 
            p=0.25,  # prev: 0.50

        ),

        v2.Normalize(

            # Pre-Trained ImageNet standards
            mean=[0.485, 0.456, 0.406],
            std=[0.229, 0.224, 0.225],

        ),

        # Try random erasing
        #   - disabeled for reference run
        #   - common after normalization (tensor-only in docs)
        # v2.RandomErasing(

        #     p=0.10,  # prev: 0.25
        #     scale=(0.01, 0.05),  # prev: (0.02, 0.15)

        # ),

    ]

)

val_transformations = v2.Compose(

    [

        v2.ToImage(),
        v2.Resize((64, 64), antialias=True),
        v2.ToDtype(torch.float32, scale=True),
        v2.Normalize(

            mean=[0.485, 0.456, 0.406], 
            std=[0.229, 0.224, 0.225]

        ),

    ]

)

test_transformations = val_transformations
```

**Model Defintion**

```python
from src.models import get_resnet_model

# Device
device = torch.device('cuda' if torch.cuda.is_available() else 'cpu')
print(f"Using Device: {device}\n")

# Define Model
model = get_resnet_model(

    model_variant=model_variant,
    num_classes=num_classes,
    use_architectural_mod=use_architectural_mod,
    stem_conv=stem_conv

)

# Load Model
model = model.to(device)  # from `get_resnet_model()` fct. in `src/models.py`

# Loss functions (independent of EMA/weight averaging
train_loss_fn = nn.CrossEntropyLoss(
    label_smoothing=label_smoothing
)
eval_loss_fn = nn.CrossEntropyLoss(label_smoothing=0.0)

# EMA: Exponential Moving Average
if weight_averaging == 'ema':

    # EMA model
    ema_model = AveragedModel(

        model, 
        multi_avg_fn=get_ema_multi_avg_fn(ema_decay),

        # For BatchNorm-heavy models such as ResNet18:
        use_buffers=True,  # average both params AND buffers (defaults to `use_buffers=False`)
    
    )

elif weight_averaging == 'none':  # default

    pass

else:

    raise ValueError(f"Unknown `weight_averaging` option: {weight_averaging}")


# Learning Rate

if lr_approach == 'fixed':

    print(f"Using Training Approach: {lr_approach}")

    optimizer = torch.optim.Adam(model.parameters(), lr=learning_rate)  # Default Adam optimizer

elif lr_approach == 'sota':

    print(f"Using Training Approach: {lr_approach}")

    # backbone_params = [param for name, param in model.named_parameters() if "fc" not in name and param.requires_grad]
    # classifier_params = [param for name, param in model.named_parameters() if "fc" in name and param.requires_grad]

    # More precise:
    backbone_params = []
    classifier_params = []

    for name, param in model.named_parameters():

        if not param.requires_grad:

            continue

        if "fc" in name or "classifier" in name:

            classifier_params.append(param)

        else:

            backbone_params.append(param)

    optimizer = torch.optim.AdamW(  # Improved AdamW optimizer: Recommended for transfer learning
        
        [  
            
            {'params': backbone_params, 'lr': backbone_param_lr}, # Lower LR for the backbone
            {'params': classifier_params, 'lr': classifier_param_lr},  # Higher LR for the new classifier
        
        ], weight_decay=weight_decay
    )

    scheduler = OneCycleLR(

        optimizer,
        max_lr=max_lr, # Maximum learning rates (backbone, classifier)
        steps_per_epoch=int(len(train_dataloader)),
        epochs=num_epochs,
        pct_start=pct_start,  # "warm up" phase
        div_factor=div_factor,  # Initial LR (max_lr/div_factor)
        final_div_factor=final_div_factor,  # Final LR
        anneal_strategy=anneal_strategy,  # Cosine annealing smoother (default: 'linear')
    
    )

else:
    
    raise ValueError(f"Unknown `lr_approach` option: {lr_approach}")
```

**Training**

```python
# Pre-Allocation

train_losses = []  # Smoothed Cross-Entropy: training uses label smoothing by default

val_losses_smoothed = []  # Smoothed Cross-Entropy: validation with same loss defintion used for training
val_losses_hard = []  # Default Cross-Entropy: for diagnostics -> actural class probabilities

best_val_loss_hard = float('inf')
delta_loss_min = 1e-4

train_accuracies = []
val_accuracies = []

best_val_accuracy = -float('inf')
delta_acc_min = 1e-6

epochs_list = []
epoch_durations = []

# LR tracking
lr_history = []

# Enable AMP usage based on employed device
use_cuda_amp = (device.type == 'cuda')

# Initialize gradient scaler
scaler = GradScaler(enabled=use_cuda_amp)

# Training loop
for epoch in range(num_epochs):

    print(f"epoch {epoch + 1}\n-------------------------------")

    train_size = len(train_dataloader.dataset)


    ### Training ###

    model.train()  # Initialize "Training Mode"

    t_epoch_start = time.time()

    cml_train_loss = 0.0  # Cumulation for subsequent mean calcualation
    loss_prev = 0.0

    train_batches = 0
    train_total = 0
    train_correct = 0

    for batch, (img, lbl) in enumerate(train_dataloader):

        # Training on device (GPU/CUDA if available)
        img = img.to(device, non_blocking=True)
        lbl = lbl.to(device, non_blocking=True)

        optimizer.zero_grad(set_to_none=True)  # None instead of zero slightly more efficient

        # Forward pass
        with autocast(  # Mixed Precision

            device_type=device.type,  # Dynamic device choice
            dtype=torch.float16,
            enabled=use_cuda_amp,  # Prevents CPU-device float16 autocast if no CUDA available

        ):  

            p = model(img)  # Pass forward

            # Prediction loss
            loss = train_loss_fn(p, lbl)

        # Backprop
        # loss.backward()  # default
        scaler.scale(loss).backward()  # Mixed Precision scaled

        # Gradient Clipping: Enhance loss stability (high learning rate phase)
        #   - Experiments show minimally better validation accuracy without clipping -> omit for simplicty
        if lr_approach == 'sota':

            scaler.unscale_(optimizer)  # Unscale gradients before clipping
            nn.utils.clip_grad_norm_(model.parameters(), max_norm=1.0) 

        # optimizer.step()
        scaler.step(optimizer)

        # Update
        scaler.update()

        # Record learning rates used for the completed step (before scheduler update)
        current_lrs = [param_group['lr'] for param_group in optimizer.param_groups]
        lr_history.append(current_lrs)

        # Update LR if using SOTA approach
        if lr_approach == 'sota':

            scheduler.step()

        # Exponential Moving Average
        #   - `ema_start_epoch`: delayed start (hardcoded) -> first few epochs noisy 
        if weight_averaging == 'ema' and (epoch >= ema_start_epoch):

            ema_model.update_parameters(model)

        # Accumulate Loss
        cml_train_loss += loss.item()
        train_batches += 1

        # Loss logging
        if (batch % 100) == 0 and batch > 0:

            loss = loss.item()

            current_sample_abs = batch * batch_size + len(img)
            current_sample_pct = abs(current_sample_abs / train_size) * 100

            # Compute incremental loss difference
            delta_loss = (loss_prev - loss) # if batch > 0 else 0.0  # Handle first batch
            loss_prev = loss

            print(f"sample: {current_sample_abs:>5d}/{train_size:>5d} ({current_sample_pct:>0.2f}%) | loss: {loss:>7f} | delta_loss: {delta_loss:>7f}")

        # Track training accuracy metrics
        _, predicted = p.max(1)
        train_total += lbl.size(0)
        train_correct += predicted.eq(lbl).sum().item()

    # Average train loss (epoch)
    avg_train_loss = cml_train_loss / train_batches
    train_losses.append(avg_train_loss)

    # Compute training accuracy
    model_train_accuracy = 100 * (train_correct / train_total)
    train_accuracies.append(model_train_accuracy)


    ### Validation ###

    # Use EMA only if exists AND synchronized with trained model at least once
    ema_active = (weight_averaging == 'ema') and int(ema_model.n_averaged.item()) > 0

    # Initialize model "evaluation mode"
    #   - includes EMA conditional 
    eval_model = ema_model if ema_active else model 
    eval_model.eval()

    val_size = len(val_dataloader.dataset)
    batch_count = len(val_dataloader)

    cml_val_loss_hard = 0.0
    cml_val_loss_smoothed = 0.0

    correct_val_predictions = 0

    with torch.no_grad():  # Avoid gradient unnecessary computation in eval mode

        for img, lbl in val_dataloader:

            img = img.to(device, non_blocking=True)
            lbl = lbl.to(device, non_blocking=True)

            p = eval_model(img)

            loss_hard = eval_loss_fn(p, lbl)
            loss_smoothed = train_loss_fn(p, lbl)

            cml_val_loss_hard += loss_hard.item()
            cml_val_loss_smoothed += loss_smoothed.item()

            correct_val_predictions += (p.argmax(1) == lbl).type(torch.float).sum().item()

    avg_val_loss_hard = cml_val_loss_hard / batch_count
    avg_val_loss_smoothed = cml_val_loss_smoothed / batch_count

    val_losses_hard.append(avg_val_loss_hard)
    val_losses_smoothed.append(avg_val_loss_smoothed)

    model_val_accuracy = 100 * (correct_val_predictions / val_size)
    val_accuracies.append(model_val_accuracy)

    epochs_list.append(epoch + 1)

    t_epoch_end = time.time()
    t_epoch_duration = t_epoch_end - t_epoch_start

    epoch_durations.append(t_epoch_duration)

    print(f"\n")
    print(f"mean train loss: {avg_train_loss:>8f} | train accuracy: {model_train_accuracy:>0.2f}%")
    print(f"mean val loss hard: {avg_val_loss_hard:>8f} | val accuracy: {model_val_accuracy:>0.2f}%")
    print(f"mean val loss smoothed: {avg_val_loss_smoothed:>8f}\n")

    ### Early Stopping ###

    accuracy_improved = model_val_accuracy > (best_val_accuracy + delta_acc_min)

    loss_tiebreak_improved = (

        abs(model_val_accuracy - best_val_accuracy) <= delta_acc_min
        and avg_val_loss_hard < (best_val_loss_hard - delta_loss_min)

    )

    # Select by accuracy first, then by "hard" Cross-Entropy loss
    if accuracy_improved or loss_tiebreak_improved:

        # Reset counter
        recovery_iter_count = 0  

        best_val_accuracy  = model_val_accuracy
        best_val_loss_hard = avg_val_loss_hard

        # State dictionary
        if ema_active == True:

            active_state_dict = ema_model.module.state_dict()

            ckpt_model_type = 'ema'

        else:

            active_state_dict= model.state_dict()

            ckpt_model_type = 'raw'

        # Save best corresponding inference checkpoint
        torch.save({

            'epoch': epoch + 1,
            'checkpoint_type': 'best_validation',
            'checkpoint_model_type': ckpt_model_type,

            'model_state_dict': active_state_dict,  # Assign correct weights (EMA or raw)

            'val_accuracy': model_val_accuracy,
            'val_loss_hard': avg_val_loss_hard,
            'val_loss_smoothed': avg_val_loss_smoothed,

            'ema_decay': ema_decay if (weight_averaging == 'ema') else None,
            'ema_n_averaged': int(ema_model.n_averaged.item()) if (weight_averaging == 'ema') else 0,

            'selection_rule': 'best val accuracy, tie-break by unsmoothed CE',

        }, model_results_dir / 'best_inference.pth')

        print(

            f"new best: saved {ckpt_model_type} weight checkpoint | "
            f"val acc: {model_val_accuracy:.2f}% | "
            f"hard val loss: {avg_val_loss_hard:.4f} | "
            f"smoothed val loss: {avg_val_loss_smoothed:.4f}\n"

        )

    else:

        # Increase counter
        recovery_iter_count += 1  

        if (recovery_iter_count >= recovery_opp) and (use_early_stopping == True):

                print(f"(!) Early Stopping: No improvement in validation accuracy / hard loss for {recovery_opp} consecutive epochs. Terminating training.\n")

                break

epochs_completed = len(epochs_list)
total_training_time = sum(epoch_durations)

print(f"Training Done! Total training duration for {epochs_completed} epochs: {total_training_time:>0.1f}s")
```

```python
# Use EMA only if exists AND synchronized with trained model at least once
ema_active = (weight_averaging == 'ema') and int(ema_model.n_averaged.item()) > 0

# State dictionary
if ema_active == True:

    final_state_dict  = ema_model.module.state_dict()

    final_ckpt_model_type = 'ema'

else:

    final_state_dict = model.state_dict()

    final_ckpt_model_type = 'raw'

# Metrics
final_epoch = epochs_list[-1]

final_train_loss = train_losses[-1]
final_val_loss_hard = val_losses_hard[-1]
final_val_loss_smoothed = val_losses_smoothed[-1]

final_train_accuracy = train_accuracies[-1]
final_val_accuracy = val_accuracies[-1]

# Final model inference checkpoint
model_checkpoint = {

    'epoch': final_epoch,

    'checkpoint_type': 'final_epoch',
    'checkpoint_model_type': final_ckpt_model_type,

    'model_state_dict': final_state_dict,

    'train_losses': train_losses,
    'val_losses_hard': val_losses_hard,
    'val_losses_smoothed': val_losses_smoothed,

    'train_accuracies': train_accuracies,
    'val_accuracies': val_accuracies,

    # Final metrics
    'final_train_loss': final_train_loss,
    'final_val_loss_hard': final_val_loss_hard,
    'final_val_loss_smoothed': final_val_loss_smoothed,

    'final_train_accuracy': final_train_accuracy,
    'final_val_accuracy': final_val_accuracy,

    # EMA metadata
    'weight_averaging': weight_averaging,
    'ema_active': ema_active,
    'ema_decay': ema_decay if weight_averaging == 'ema' else None,
    'ema_n_averaged': int(ema_model.n_averaged.item()) if weight_averaging == 'ema' else 0,

    # Hyperparam.
    'hyperparameters': {

        'model_variant': model_variant,
        'architectural_mod': use_architectural_mod,
        'stem_conv': stem_conv,

        'lr_approach': lr_approach,
        'num_epochs': num_epochs,
        'epochs_completed': epochs_completed,

        'batch_size': batch_size,
        'num_workers': num_workers,

        'backbone_lr': backbone_param_lr if lr_approach == 'sota' else None,
        'classifier_lr': classifier_param_lr if lr_approach == 'sota' else None,
        'learning_rate': learning_rate if lr_approach == 'fixed' else None,

        'weight_decay': weight_decay if lr_approach == 'sota' else None,
        'pct_start': pct_start if lr_approach == 'sota' else None,
        'div_factor': div_factor if lr_approach == 'sota' else None,
        'final_div_factor': final_div_factor if lr_approach == 'sota' else None,
        'anneal_strategy': anneal_strategy if lr_approach == 'sota' else None,

        'label_smoothing': label_smoothing,
        'rng_seed': rng_seed,

    }

}

# Save inference checkpoint
torch.save(model_checkpoint, model_results_dir / 'last_inference.pth')

# Final model "resume" checkpoint
resume_checkpoint = {

    'epoch': epochs_completed,
    'raw_model_state_dict': model.state_dict(),
    'optimizer_state_dict': optimizer.state_dict(),
    'scheduler_state_dict': scheduler.state_dict() if (lr_approach == 'sota') else None,
    'scaler_state_dict': scaler.state_dict(),
    'ema_state_dict': ema_model.state_dict() if (weight_averaging == 'ema') else None,

    'training_history': {

        'train_losses': train_losses,
        'val_losses_hard': val_losses_hard,
        'val_losses_smoothed': val_losses_smoothed,
        'train_accuracies': train_accuracies,
        'val_accuracies': val_accuracies,
        'lr_history': lr_history,
        'epoch_durations': epoch_durations,
        
    },

    'hyperparameters': {

        'model_variant': model_variant,
        'architectural_mod': use_architectural_mod,
        'stem_conv': stem_conv,

        'lr_approach': lr_approach,
        'num_epochs': num_epochs,
        'epochs_completed': epochs_completed,

        'batch_size': batch_size,
        'num_workers': num_workers,

        'backbone_lr': backbone_param_lr if lr_approach == 'sota' else None,
        'classifier_lr': classifier_param_lr if lr_approach == 'sota' else None,
        'learning_rate': learning_rate if lr_approach == 'fixed' else None,

        'weight_decay': weight_decay if lr_approach == 'sota' else None,
        'pct_start': pct_start if lr_approach == 'sota' else None,
        'div_factor': div_factor if lr_approach == 'sota' else None,
        'final_div_factor': final_div_factor if lr_approach == 'sota' else None,
        'anneal_strategy': anneal_strategy if lr_approach == 'sota' else None,

        'label_smoothing': label_smoothing,
        'rng_seed': rng_seed,

    }

}

# Save resume checkpoint
torch.save(resume_checkpoint, model_results_dir / 'last_resume.pth')

print(

    f"Saved final {final_ckpt_model_type} checkpoint after {epochs_completed} completed epochs | "
    f"val acc: {final_val_accuracy:.2f}% | "
    f"hard val loss: {final_val_loss_hard:.4f} | "
    f"smoothed val loss: {final_val_loss_smoothed:.4f}"

)
```
